# Rayuela Mortal: Stochastic Simulation Analysis
This notebook explores the dynamics of the 'Rayuela Mortal' game using NumPy and Pandas for optimized simulation and analysis.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from main import run_experiment_1, run_experiment_2, run_experiment_3

sns.set_theme(style="whitegrid")
```

## Experiment 1: Board Scaling
How does the size of the board affect the expected number of turns? Theory suggests an exponential or high-order polynomial growth due to the reset mechanism.

```python
df1 = run_experiment_1(max_size=25, iterations=100)
plt.figure(figsize=(10, 6))
sns.lineplot(data=df1, x='Board Size', y='Avg Duration', marker='o')
plt.title("Expected Game Duration vs Board Size")
plt.show()
```

## Experiment 2: The Rock-Paper-Scissors Paradox
Does biasing one's strategy in a zero-sum game with perfect information (assuming the opponent stays random) yield any advantage?

```python
df2 = run_experiment_2(iterations=200)
plt.figure(figsize=(10, 6))
sns.barplot(data=df2, x='P(Rock)_A', y='Win Rate A', palette='viridis')
plt.axhline(0.5, ls='--', color='red')
plt.title("Effect of Biased RPS Strategy on Win Rate")
plt.show()
```

## Experiment 3: Kinetic Advantage
The impact of the probability of advancing on the final outcome.

```python
df3 = run_experiment_3(iterations=200)
df3_melted = df3.melt('P(Advance)_A', var_name='Team', value_name='Win Rate')
plt.figure(figsize=(10, 6))
sns.lineplot(data=df3_melted, x='P(Advance)_A', y='Win Rate', hue='Team', marker='s')
plt.title("Win Rate as a Function of Movement Probability")
plt.show()
```
